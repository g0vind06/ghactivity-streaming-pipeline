```python
import boto3
import json
import gzip
import time
import glob
import os
```

```python
os.environ.setdefault('AWS_DEFAULT_PROFILE', 'Govind')
os.environ.setdefault('AWS_DEFAULT_REGION', 'us-east-1')
```

```python
kinesis=boto3.client('kinesis')
```

```python
stream_name = 'ghactivity-stream'
```

```python
#producer function

def send_events(file_pattern):
    files = glob.glob(file_pattern)    #returns list of files matching the pattern

    for file in files:
        print(f"Processing {file}...")

        with gzip.open(file, 'rt') as f:
            for line in f:
                event=json.loads(line)
                kinesis.put_record(
                StreamName=stream_name,
                Data=json.dumps(event),
                PartitionKey=event.get('type', 'default')
                )
                time.sleep(0.1)                 
```

```python
send_events('/home/gsheshmanitiwari/Projects/ghactivity/data/*.json.gz')
```
